```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Line3DCollection
```

```python
df=pd.read_csv("trajectory.csv")
N=10
#ax.plot(df.x1,df.y1,df.z1,label="Sun" )
#ax.plot(df.x2,df.y2,df.z2,label="Earth")
xs = df.iloc[:, 1:1+N].to_numpy()
ys = df.iloc[:, 1+N:1+2*N].to_numpy()
zs = df.iloc[:, 1+2*N:1+3*N].to_numpy()

ax = plt.figure().add_subplot(projection='3d')
colors = plt.cm.viridis(np.linspace(0, 1, N))

# transpose so each row becomes one trajectory
ax.plot(xs.T, ys.T, zs.T)

ax.set_xlim3d(0,4)
ax.set_ylim3d(0,4)
ax.set_zlim3d(0,4)
ax.view_init(90,180)
```

```python
plt.plot(xs,ys)
plt.ylim(0,4)
plt.xlim(0,4)
```

```python
Nt, N = xs.shape  # xs shape (Nt, Nparticles)

# Build segments array: (Nparticles, Nt, 3)
segments = np.stack((xs.T, ys.T, zs.T), axis=2)

# Create collection
lc = Line3DCollection(segments)

# Apply colormap
colors = plt.cm.viridis(np.linspace(0, 1, N))
lc.set_colors(colors)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.add_collection3d(lc)

ax.set_xlim(0,4)
ax.set_ylim(0,4)
ax.set_zlim(0,4)

ax.view_init(70, -90)

plt.show()
```
